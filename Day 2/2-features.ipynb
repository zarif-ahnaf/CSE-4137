```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
```

```python
df = pd.read_excel("./data.ods",engine='odf')
df.head()
```

```python
X = df.iloc[:,:-1]
y=df.iloc[:,-1]
```

```python
X
```

```python
y
```

```python
plt.figure(figsize=(4,3))
plt.scatter(X,y)
plt.show()
```

```python
lr = LinearRegression()
lr.fit(X,y)
```

```python
m = lr.coef_
```

```python
b = lr.intercept_
```

```python
x1 = np.linspace(60,200,20)
```

```python
x1
```

```python
y1 = m*x1+b
```

```python
y1
```

```python
plt.figure(figsize=(4,3))
plt.scatter(X,y)
plt.plot(x1,y1)
plt.show()
```

```python
y_pred = lr.predict(X)
y_pred
```

```python
y
```

```python
mae = mean_absolute_error(y,y_pred)
mae
```

```python
mse = mean_squared_error(y,y_pred)
mse
```

```python
r2s = r2_score(y,y_pred)
print(f"Score={r2s*100}")
```

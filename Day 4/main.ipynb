```python
kx  = [ ]
wcss = [ ]
for k in range(1,10):
    km = KMeans(n_clusters=k)
    y_pred = km.fit_predict(df)
    kx.append(k)
    wcss.append(km.inertia_)
    print(f'k={k}, WCSS={km.inertia_}')
```

```python
plt.figure(figsize=(4,3))
plt.plot(kx,wcss)
plt.xlabel("K")
plt.ylabel("WCSS")
```
